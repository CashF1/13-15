```python
#15.1

import multiprocessing
import random
import time
from datetime import datetime

def process():
    time.sleep(random.random())
    print(datetime.now())
    
if __name__ == "__main__":
    processes = []

    for i in range(3):
        p = multiprocessing.Process(target=process)
        processes.append(p)
        p.start()
    
    for p in processes:
        p.join()
        
```
